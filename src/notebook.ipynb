```python
import pandas as pd

# Load CSV file into a DataFrame
file_path = '/Users/manhtd/Documents/Projects/website-crawler/output/websites.csv'  # Replace with your CSV file path
df = pd.read_csv(file_path)

# Display the first few rows of the DataFrame
print(df.head())
```

```python
from urllib.parse import urlparse
from collections import Counter

def get_website_domains(df):
    # Extract domain names from the websites column
    domains = df['websites'].apply(lambda url: urlparse(url).netloc)
    
    # Count the occurrences of each domain
    domain_counts = Counter(domains)
    
    # Convert the result to a DataFrame
    result_df = pd.DataFrame(domain_counts.items(), columns=['Domain', 'Count']).sort_values(by='Count', ascending=False)
    return result_df

# Get the domain counts
domain_counts_df = get_website_domains(df)

# Display the result
domain_counts_df.to_csv('domain_counts.csv', index=False)
```
